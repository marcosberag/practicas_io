```python
import numpy as np
from tensorflow.keras.datasets import mnist
from sklearn.svm import SVC

# fijamos la semilla (SVC con estos parámetros no usa aleatoriedad, pero así
# el resto del cuaderno es reproducible)
np.random.seed(42)

# cargamos MNIST: 60.000 imágenes de entrenamiento y 10.000 de test
(x_train, y_train), (x_test, y_test) = mnist.load_data()

# normalizamos los píxeles de 0-255 a 0-1
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

# aplanamos cada imagen 28x28 en un vector de 784 números (lo que espera la SVM)
x_train_svm = x_train.reshape(len(x_train), -1)
x_test_svm = x_test.reshape(len(x_test), -1)

# SVM con núcleo gaussiano (RBF) y los hiperparámetros de los apuntes
svm = SVC(kernel="rbf", C=10, gamma=0.01)

# entrenamos resolviendo el problema dual
svm.fit(x_train_svm, y_train)

# Acierto en el conjunto de test
acierto = svm.score(x_test_svm, y_test)
print("Acierto en test:", acierto)

# Número de vectores soporte (puntos con lambda > 0)
print("Vectores soporte:", svm.support_vectors_.shape[0])
```

```python
import joblib

# Guardamos la SVM entrenada en un archivo
joblib.dump(svm, "svm_mnist.joblib")
```
